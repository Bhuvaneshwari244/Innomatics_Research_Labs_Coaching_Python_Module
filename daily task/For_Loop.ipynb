## 1. Lowercase Alphabets Printer

## 7. List Reverser

Problem Statement:

 Write a program that reverses a given list using a for loop.

Test Cases:

Input: List = [10, 20, 30, 40]

 Output: [40, 30, 20, 10]


Input: List = [1, 2, 3]

 Output: [3, 2, 1]



```python
user_input = input("Enter list elements separated by spaces: ")
lst = user_input.split()
reversed_lst = []
# len(lst) - 1 is the last index; -1 is the stop index (not included); -1 is the step
for i in range(len(lst) - 1, -1, -1):
    reversed_lst.append(lst[i])

print("Output:", reversed_lst)
```

## 11. Perfect Square Checker

 Write a program to check whether a given number is a perfect square, using a for loop.

Test Cases:

Input: Number = 49

 Output: Perfect Square


Input: Number = 50

 Output: Not a Perfect Square



```python
num = int(input('Enter number: '))

for i in range(1, num+1):
  if i * i == num:
    print('perfect square')
    break

if i * i != num:
      print('not perfect square')
```

## 12. Prime Numbers up to N

Problem Statement:

 Write a program that prints all prime numbers up to a given number N, using a for loop.

Test Cases:

Input: N = 20

 Output: 2 3 5 7 11 13 17 19


Input: N = 10

 Output: 2 3 5 7


```python
N = int(input("Enter N: "))

for num in range(2, N + 1):
    for i in range(2, num):
        if num % i == 0:
            break
    else:
        print(num, end=" ")
print()
```

Input: N = 7

 Output: 0 1 1 2 3 5 8


Input: N = 5

 Output: 0 1 1 2 3



```python
N = int(input("Enter N: "))

a, b = 0, 1

for i in range(N):
    print(a, end=" ")
    a, b = b, a + b
print()
```

## 14. Armstrong Number Checker

Problem Statement:

 Write a program to check whether a given number is an Armstrong number, using a for loop.

Test Cases:

Input: Number = 153

 Output: Armstrong


Input: Number = 123

 Output: Not Armstrong



```python
num = input("Enter Number: ")
p = len(num)
total = 0

for x in num:
    total += int(x) ** p
if total == int(num):
    print("Armstrong")
else:
    print("Not Armstrong")
```

```python
num_str = input("Enter Number: ")
num_digits = len(num_str)
original_num = int(num_str)

armstrong_sum = 0
for digit in num_str:
    armstrong_sum += int(digit) ** num_digits

if armstrong_sum == original_num:
    print("Armstrong")
else:
    print("Not Armstrong")
```

## 15. Frequency Counter

Problem Statement:

 Write a program that counts how many times a given value appears in a list, using a for loop.

Test Cases:

Input: List = [2, 3, 2, 5, 2], Find = 2

 Output: Count = 3


Input: List = [1, 1, 2, 3], Find = 1

 Output: Count = 2




```python
items = input("Enter elements: ").split()
find_val = input("Enter value to find: ")

count = 0
for x in items:
    if x.strip("[],") == find_val:
        count += 1
print("Count =", count)
```

## 17. Common Elements Finder

Problem Statement:

 Write a program that finds and prints common elements between two lists using a for loop.

Test Cases:

Input: List1 = [1, 2, 3], List2 = [2, 3, 4]

 Output: [2, 3]


Input: List1 = [5, 6], List2 = [7, 8]

 Output: []



```python
list1 = input("Enter List1: ").split()
list2 = input("Enter List2: ").split()

common = []

for x in list1:
    if x in list2:
        num = int(x)
        if num not in common:
            common.append(num)

print(common)
```

```python
list1 = [int(x) for x in input("Enter List1 elements: ").split()]
list2 = [int(x) for x in input("Enter List2 elements: ").split()]

common_elements = []

for item in list1:
    if item in list2 and item not in common_elements:
        common_elements.append(item)

print(common_elements)
```

## 19. Palindrome Word Checker

Problem Statement:

 Write a program to check whether a given word is a palindrome, using a for loop.

Test Cases:

Input: Word = "level"

 Output: Palindrome


Input: Word = "python"

 Output: Not Palindrome



```python
word = input("Enter Word: ")

reversed_word = ""
for i in range(len(word) - 1, -1, -1):
    reversed_word += word[i]

if word == reversed_word:
    print("Palindrome")
else:
    print("Not Palindrome")
```

## 20. Greatest Number in a List

Problem Statement:

 Write a program that finds the greatest number in a list using a for loop.

Test Cases:

Input: List = [10, 35, 7, 99]

 Output: Greatest = 99


Input: List = [1, 2, 3]

 Output: Greatest = 3




```python
user_input = input("Enter list elements separated by space: ")
lst = user_input.split()

# Set max_val to the first element in your list
max_val = int(lst[0])

for num in lst:
    # Convert each element to an integer to compare them correctly
    if int(num) > max_val:
        max_val = int(num)

print("Greatest =", max_val)
```
