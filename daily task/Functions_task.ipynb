### 1. Why Return?

 Write a function add(a, b) that returns the sum of two numbers. Call it and print the returned value. Why is return used instead of print inside the function?

Answer:

return sends the calculated value back to the place where the function was called. This allows the output to be stored in a variable, passed to other functions, or used in further calculations. print() only displays text on the screen and returns None, meaning the result cannot be reused in code logic

```python
def add(a,b):
  return a+b
add(5,2)
##print just displays the output where return will send the value back to the function whenever called
```

In Python, if a function does not contain an explicit return statement, it automatically returns None by default upon completing execution.

## 3. Positional Arguments

Question:

Write a function greet(first, last) that prints "Hello, first last!". Call it with different names, maintaining order.

Answer:

Positional arguments match values to parameters based strictly on the order in which they are passed.

```python
def greet(first,last):
  print(first,last)
greet('hello','first last!')
```

## 4. Keyword Arguments

Question:

Using the same greet function, call it using greet(last="Smith", first="John"). What happens if the order changes?

Answer:

When using keyword arguments (parameter_name=value), Python assigns values based on parameter names rather than their order. Changing the order in the function call does not affect the output.

```python
def greet(first,last):
  print(first,last)
greet(last ='smith',first = 'john')
```

## 6. *Using *args

Question:

Write a function multiply(*nums) that multiplies all numbers passed to it. Call it with 2, 3, 4 and print the result.

Answer:

*args allows a function to accept an arbitrary number of positional arguments, which are gathered inside the function as a tuple.

```python
def multiply(*nums):
  res = 1
  for i in nums:
    res*=i
  return res
multiply(2,3,4)
```

## 7. Using kwargs

Question:

Write a function student_info(**details) that prints each key-value pair passed (e.g., name="Anu", age=20).

Answer:

**kwargs allows a function to accept any number of keyword arguments, which are collected into a dictionary.

```python
def student_info(**details):
    for key, value in details.items():
        print(key + ":", value)

student_info(name="Anu", age=20, course="Python")
```

## 8. Mixing Positional and Keyword Arguments

Question:

Create a function order(item, price, quantity=1) and call it with a mix of positional and keyword arguments.

Answer:

When mixing positional and keyword arguments, positional arguments must always precede keyword arguments.

```python
def order(item, price, quantity=1):
    print("Item:", item, "| Price:", price, "| Quantity:", quantity)

order("Laptop", price=55000, quantity=2)
```

## 9. Returning Multiple Values

Question:

Write a function calculate(a, b) that returns both the sum and product of a and b. How do you capture both values?

Answer:

When a function returns multiple values separated by commas, Python packs them into a tuple. You can capture them using multiple assignment (tuple unpacking).

```python
def calculate(a, b):
    return a + b, a * b
sum_val, product_val = calculate(5, 3)
print("Sum:", sum_val, "| Product:", product_val)
```

## 10. Function that Does Nothing

Question:

Define a function placeholder() with pass. What happens when you call it? Why is it sometimes used in code?

Answer:

Calling it executes without errors and returns None. The pass statement acts as a placeholder when syntactic structure requires a block of code, but no action is needed yet (useful during planning or drafting code structure).

```python
def placeholder():
    pass

result = placeholder()
print("Result of placeholder():", result)
```

Problem: Write a function that takes the price of an item and GST percentage, calculates the final amount, and returns it. Store the returned value in a variable and print the bill amount.

6. Shopping Cart Total (*args)

Problem: Create a function that accepts any number of item prices and returns the total bill amount. Test it with different numbers of prices.
