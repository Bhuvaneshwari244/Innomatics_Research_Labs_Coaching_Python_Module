1. Input Validation - Age Check

Write a while loop that asks the user to enter their age. If the age is less than 0 or more than 120, ask again until a valid age is entered.

Example: -5 → 130 → 25

Valid input: 25

```python


while True:
    age = int(input())
    if(0 <= age <= 120):
        print("valid input")
        break
    else:
        continue
```

2. Repeat Until Vowel

Keep asking the user to enter a character until they enter a vowel (a, e, i, o, u).

Input: k → b → o

Output: "Vowel detected"

```python
while True:
    char = input()
    if(char.lower() in ['a','e','i','o','u',]):
        print('vowel detected')
        break
    else:
        continue
```

3. Print Squares of Numbers from 1 to n

Ask the user for a number n, and use a while loop to print squares of numbers from 1 to n.

Input: 4

Output: 1 4 9 16

```python
n = int(input())
i = 1
while i<=n:
    print(i**2,end=" ")
    i+=1
```

3. Countdown Timer in Seconds

Ask for a time in seconds and simulate a countdown using a while loop.

Input: 5

Output: 5 4 3 2 1 Time's up!

```python
time = int(input())
while time >= 1:
    print(time,end=" ")
    time -= 1
print("Time's up!")
```

4. Keep Asking for a Username Until Length ≥ 6

Use a while loop to ask the user for a username until they enter at least 6 characters.

Input: abc → bcda → abcded

Output: Username accepted

```python
name = input()
while True:
    if(len(name) >= 6):
        print('username accepted')
        break
    else:
        name = input()
```

5. Find the First Number Divisible by Both 3 and 7 (from user input)

Ask the user for a starting number and keep incrementing it using a loop until a number divisible by both 3 and 7 is found.

Input: 10

Output: 21

```python
num = int(input())
while num%3 !=0 or num%7 !=0:
    num += 1
print(num)
```

Create a while loop that keeps accepting names from the user. Stop the loop if they enter a blank string.

8. Sum All Odd Numbers Till n

Ask the user for a number n. Use a while loop to sum all odd numbers between 1 and n.

Input: 10

Output: 25

```python
n = int(input())
s = 0
i = 1
while i<=n:
    s += i
    i += 2
print(s)
```

12. Keep Accepting Numbers, Count How Many Were Even

Stop the loop when the user enters -1.

Input: 2 → 3 → 4 → 7 → -1

Output: Even numbers: 2

```python
num = int(input())
count = 0
while num != -1:
    if(num % 2 == 0):
        count += 1
    num = int(input())
print("even numbers: ", count)
```

14. Use While Loop to Remove All Vowels from a Word

Input: "pythonista"

Output: "pythnst"

```python
inp = input()
i = 0
while i < len(inp):
    if(inp[i].lower() not in ['a','e','i','o','u']):
        print(inp[i], end = "")
    i += 1
```

15. Enter Items into a List Until the User Types "stop", Then Print the List

Input: apple → banana → stop

Output: ['apple', 'banana']

```python
items = []
item = input()
while item.lower() != 'stop':
    items.append(item)
    item = input()
print(items)
```
