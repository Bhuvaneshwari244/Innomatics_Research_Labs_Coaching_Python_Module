2. Create a program where a function area() calculates the area of both a circle and a rectangle (demonstrating polymorphism).

```python
def area(shape, a, b=0):
    if shape == "circle":
        print("Circle Area:", 3.14 * a * a)
    elif shape == "rectangle":
        print("Rectangle Area:", a * b)

area("circle", 5)
area("rectangle", 4, 6)
```

3. Write a base class Animal with a method sound(). Override sound() in subclasses Dog and Cat to demonstrate polymorphism.


```python
class Animal:
    def sound(self):
        print("Some animal sound")

class Dog(Animal):
    def sound(self):
        print("Woof Woof!")

class Cat(Animal):
    def sound(self):
        print("Meow Meow!")

for pet in [Dog(), Cat()]:
    pet.sound()
```

#Abstraction (3 Questions)

4. What is abstraction in Python? Show how it is implemented using the abc module with an example.


```python
from abc import ABC, abstractmethod

class Demo(ABC):
    @abstractmethod
    def display(self):
        pass

class ChildDemo(Demo):
    def display(self):
        print("Abstract method implemented!")

c = ChildDemo()
c.display()
```

5. Create an abstract class BankAccount with abstract methods deposit() and withdraw(). Implement them in a subclass SavingsAccount.



```python
from abc import ABC, abstractmethod

class BankAccount(ABC):
    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass

class SavingsAccount(BankAccount):
    def deposit(self, amount):
        print(f"Deposited ${amount}")

    def withdraw(self, amount):
        print(f"Withdrew ${amount}")

s = SavingsAccount()
s.deposit(500)
s.withdraw(200)
```

6. Create an abstract class Vehicle with a method start(). Implement it in subclasses Car and Bike, showing different ways of starting each vehicle.


```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car starts with a Key/Button push.")

class Bike(Vehicle):
    def start(self):
        print("Bike starts with a Kick/Self start.")

Car().start()
Bike().start()
```

#Encapsulation (4 Questions)

7. What is public, protected, and private access in Python? Demonstrate all three using an Employee class.


```python
class Employee:
    def __init__(self, name, dept, salary):
        self.name = name
        self._department = dept
        self.__salary = salary

    def show_private(self):
        print("Private Salary:", self.__salary)

e = Employee("John", "IT", 50000)
print("Public Name:", e.name)
print("Protected Dept:", e._department)
e.show_private()
```

8. Write a Student class where marks are kept private and can only be accessed/modified using getter and setter methods.


```python
class Student:
    def __init__(self):
        self.__marks = 0

    def set_marks(self, marks):
        if marks >= 0 and marks <= 100:
            self.__marks = marks
        else:
            print("Invalid Marks!")

    def get_marks(self):
        return self.__marks

s = Student()
s.set_marks(85)
print("Student Marks:", s.get_marks())
```

9. Create a BankAccount class where the account balance is private. Provide methods for deposit, withdraw, and check balance to demonstrate encapsulation.


```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient Balance!")

    def check_balance(self):
        print("Current Balance:", self.__balance)

acc = BankAccount(1000)
acc.deposit(500)
acc.withdraw(200)
acc.check_balance()
```

10. Write a Car class where speed is private. Provide methods to accelerate, brake, and check speed.


```python
class Car:
    def __init__(self):
        self.__speed = 0

    def accelerate(self, increment):
        self.__speed += increment

    def brake(self, decrement):
        self.__speed = max(0, self.__speed - decrement)

    def check_speed(self):
        print("Current Speed:", self.__speed, "km/h")

my_car = Car()
my_car.accelerate(50)
my_car.check_speed()
my_car.brake(20)
my_car.check_speed()
```
