18. What is the difference between sort() and reverse()?

- `sort()` organizes items based on their value (like alphabetical or numerical order).
- `reverse()` organizes items based on their position (flipping the sequence from back to front).

#Copying Lists

```python
a = [1, 2, 3]
a.copy()
```

```python
a = [1, 2, 3]
a1 = a.copy()
a1
```

20. What is the difference between a = b and a = b.copy()?

- `a = b` creates a shortcut. Both variables point to the exact same list. Changing a also changes b.
- `a = b.copy()` creates a clone. It makes a brand-new, independent list. Changing a leaves b untouched.
