#Tuple Questions

1) Tuple Count


- Given the tuple colors = ('red', 'blue', 'green', 'red', 'yellow', 'red'), count how many times 'red' appears.



```python
colors = ('red', 'blue', 'green', 'red', 'yellow', 'red')
colors.count('red')
```

2) Tuple Index


- Find the index of 'green' in the tuple colors = ('red', 'blue', 'green', 'yellow').




```python
colors = ('red', 'blue', 'green', 'yellow')
colors.index('green')
```

 # Set Questions

3) Set Add


- Add 'cherry' to the set: fruits = {'apple', 'banana'}


```python
fruits = {'apple', 'banana'}
fruits.add('cherry')
fruits
```

4) Set Update


- Add elements from another set {'orange', 'grape'} to fruits = {'apple'} using update().

```python
fruits = {'apple'}
fruits.update({'orange', 'grape'})
fruits
```

5) Set Discard


- Discard 'banana' from fruits = {'apple', 'banana', 'cherry'}.

```python
fruits = {'apple', 'banana', 'cherry'}
fruits.discard('banana')
fruits
```

6) Set Remove


- Remove 'apple' using remove() from fruits = {'apple', 'grape'}

```python
fruits = {'apple', 'grape'}
fruits.remove('apple')
fruits
```

7) Set Pop


- Pop any element from colors = {'red', 'green', 'blue'} and store it in a variable.

```python
colors = {'red', 'green', 'blue'}
colors2 = colors.pop()
colors2
```

8) Set Clear


- Clear all elements from numbers = {1, 2, 3, 4}.



```python
numbers = {1, 2, 3, 4}
numbers.clear()
numbers
```

9) Set Copy


- Make a copy of the set data = {1, 2, 3} and store it in a new variable.

```python
data = {1, 2, 3}
data1 = data.copy()
data1
```

10) Set Difference


- Find the difference between set1 = {1, 2, 3, 4} and set2 = {3, 4, 5, 6}.

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
set1.difference(set2)
```

11) Set Union


- Find the union of a = {'x', 'y'} and b = {'y', 'z'}.

```python
a = {'x', 'y'}
b = {'y', 'z'}
a.union(b)
```

12) Set Intersection


- Find common elements in set1 = {10, 20, 30} and set2 = {20, 40, 60}.

```python
set1 = {10, 20, 30}
set2 = {20, 40, 60}
set1.intersection(set2)
```

13) Set issubset


- Check if {1, 2} is a subset of {1, 2, 3, 4}.

```python
a = {1, 2, 3, 4}
b = {1, 2}
b.issubset(a)
```

14) Set issuperset


- Check if {1, 2, 3} is a superset of {2, 3}.

```python
c = {2, 3}
d = {1, 2, 3}
d.issuperset(c)
```

15) Set symmetric_difference


- Find elements not common in a = {1, 2, 3} and b = {3, 4, 5}.

```python
a = {1, 2, 3}
b = {3, 4, 5}
a.symmetric_difference(b)
```
