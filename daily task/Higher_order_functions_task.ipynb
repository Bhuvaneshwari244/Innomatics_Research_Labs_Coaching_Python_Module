# Step 1: Lambda Function

1. Write a function using lambda that takes a number and returns its square.


```python
square = lambda x: x ** 2

num = int(input("Enter a number: "))
print("Square:", square(num))
```

```python
num_input = input("Enter a number: ")
num = int(num_input)

sq = lambda x: x ** 2
print("Square =", sq(num))
```

2. Create a lambda function that takes two numbers and returns their sum.


```python
add_two_nums = lambda a, b: a + b

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Sum:", add_two_nums(a, b))
```

```python
a_input = input("Enter first number: ")
b_input = input("Enter second number: ")

a = float(a_input)
b = float(b_input)

ad = lambda x, y: x + y
print("Sum =", ad(a, b))
```

3. Write a lambda function that checks if a number is even.


```python
num_input = input("Enter an integer: ")
num = int(num_input)

even = lambda x: x % 2 == 0
print("Is even:", even(num))
```

```python
is_even = lambda x: x % 2 == 0

num = int(input("Enter an integer: "))
if is_even(num):
    print(num, "is Even")
else:
    print(num, "is Odd")
```

4. Create a lambda function that converts a temperature from Celsius to Fahrenheit.

```python
c_input = input("Enter temperature in Celsius: ")
celsius = float(c_input)

cel = lambda c: (c * 9/5) + 32
print("Fahrenheit =", cel(celsius))
```

```python
celsius_to_fahrenheit = lambda c: (c * 9/5) + 32

celsius = float(input("Enter temperature in Celsius: "))
print("Temperature in Fahrenheit:", celsius_to_fahrenheit(celsius))
```

5. Write a lambda function that takes a string and returns it in uppercase.


```python
text = input("Enter a string: ")

st = lambda s: s.upper()
print("Uppercase =", st(text))
```

```python
to_uppercase = lambda s: s.upper()

text = input("Enter a string: ")
print("Uppercase:", to_uppercase(text))
```

# Step 2: Higher-Order Functions

6. Given a list of numbers, write a program to double each number in the list.


```python
user_input = input("Enter numbers separated by space: ")

lst = []
for item in user_input.split():
    lst.append(int(item))

doub = list(map(lambda x: x * 2, lst))
print("Doubled list =", doub)
```

7. Given a list of numbers, write a program to keep only the numbers greater than 10.


```python
user_input = input("Enter numbers separated by space: ")

lst = []
for item in user_input.split():
    lst.append(float(item))

res = list(filter(lambda x: x > 10, lst))
print("Numbers greater than 10 =", res)
```

8. Given a list of numbers, write a program to calculate the product of all numbers in the list.


```python
from functools import reduce

user_input = input("Enter numbers separated by space: ")

lst = []
for item in user_input.split():
    lst.append(float(item))

res = reduce(lambda x, y: x * y, lst)
print("Product =", res)
```

# Step 3: Higher-Order Functions


9. Given a list of numbers, write a program to triple each number using a single expression.


```python
user_input = input("Enter numbers separated by space: ")

lst = []
for item in user_input.split():
    lst.append(float(item))

res = list(map(lambda x: x * 3, lst))
print("Tripled list =", res)
```

10. From a list of numbers, write a program to keep only the odd numbers using a single expression.

```python
user_input = input("Enter integers separated by space: ")

lst = []
for item in user_input.split():
    lst.append(int(item))

res = list(filter(lambda x: x % 2 != 0, lst))
print("Odd numbers =", res)
```

11. From a list of numbers, write a program to calculate the sum of all squares of the numbers.

```python
user_input = input("Enter numbers separated by space: ")

lst = []
for item in user_input.split():
    lst.append(float(item))

res = sum(map(lambda x: x ** 2, lst))
print("Sum of squares =", res)
```

12. Given a list of strings, write a program to find the length of each string in a single expression.

```python
words = input("Enter words separated by space: ").split()

string_lengths = list(map(len, words))
print("Length of each word:", string_lengths)
```

```python
user_input = input("Enter strings separated by space: ")
lst = user_input.split()

res = list(map(lambda x: len(x), lst))
print("String lengths =", res)
```

13. From a list of numbers, write a program to find the largest number using a single expression.


```python
from functools import reduce

user_input = input("Enter numbers separated by space: ")

lst = []
for item in user_input.split():
    lst.append(int(item))

res = reduce(lambda x, y: x if x > y else y, lst)
print("Largest number =", res)
```

14. Given a list of numbers, write a program to calculate the sum of the cubes of all even numbers.

```python
user_input = input("Enter integers separated by space: ")

lst = []
for item in user_input.split():
    lst.append(int(item))

res = sum(map(lambda x: x ** 3, filter(lambda x: x % 2 == 0, lst)))
print("Sum of cubes of even numbers =", res)
```
