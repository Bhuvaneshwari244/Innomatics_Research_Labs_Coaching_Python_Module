### Program 1.1: Stop at First Negative Number

Description

Print numbers from the list until a negative number appears.
 Once a negative number is found, stop the loop.

Input

data = [4, 7, 2, -1, 8, 10]

Output

4

 7

 2

Test Cases

Input

data = [10, 20, -5, 30]

Output

10

 20


```python
user_input = input("Enter numbers separated by space: ")

data = []
for item in user_input.split():
    data.append(int(item))

for num in data:
    if num < 0:
        break
    print(num, end=" ")
print()
```

Program 6.2: Menu Driven Program using break and continue

Create a simple menu system.

If input is invalid → skip using continue

If user enters "hello" → print greeting message

Test Cases

Output

Description

If every mark is greater than or equal to 35, print "All Passed".
 Otherwise print "Failed Subject Found".

```python
user_input = input("Enter marks separated by space: ")
marks = []
for item in user_input.split():
    marks.append(int(item))

for mark in marks:
    if mark < 35:
        print("Failed Subject Found")
        break
else:
    print("All Passed")
```

## TASK 11: while with else

Concept

The else block executes if the loop ends naturally without break.

Program 11.1: ATM PIN Verification

Description

The user has only 3 attempts to enter the correct ATM PIN.

 If entered correctly, print "Access Granted".

 Otherwise print "Card Blocked".

Test Cases

Test Case 1

Input

1111

 2222

 1234

Output

Wrong PIN

 Wrong PIN

 Access Granted


```python
correct_pin = "1234"
attempts = 0

while attempts < 3:
    pin = input("Enter ATM PIN: ")
    attempts += 1
    if pin == correct_pin:
        print("Access Granted")
        break
    else:
        print("Wrong PIN")
else:
    print("Card Blocked")
```
