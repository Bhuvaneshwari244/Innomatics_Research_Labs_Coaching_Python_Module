Q25. Find Unique Elements from Two Lists

Test Cases:

Input: [1, 2, 3], [3, 4, 5] → Output: [1, 2, 4, 5]

 Input: ['a', 'b'], ['b', 'c'] → Output: ['a', 'c']

 Input: [10], [10] → Output: []


```python
input1 = input("Enter first list elements separated by space: ")
input2 = input("Enter second list elements separated by space: ")

l1 = input1.split()
l2 = input2.split()

unique = []
for x in l1:
    if x not in l2 and x not in unique:
        unique.append(x)
for x in l2:
    if x not in l1 and x not in unique:
        unique.append(x)

print("Unique elements =", unique)
```

```python
l1 = [1, 2, 3]
l2 = [3, 4, 5]
unique = []
for x in l1 + l2:
    if (x in l1) ^ (x in l2):
        unique.append(x)
print(unique)  # Output: [1, 2, 4, 5]
```

Q31. Flatten a Nested List

Test Cases:

Input: [[1, 2], [3, 4], [5]] → Output: [1, 2, 3, 4, 5]

 Input: [['a', 'b'], ['c']] → Output: ['a', 'b', 'c']

 Input: [[10], [20, 30]] → Output: [10, 20, 30]


```python
rows = int(input("How many sublists do you want to enter? "))
nested_lst = []
for i in range(rows):
    sub = input(f"Enter elements for sublist {i+1} separated by space: ").split()
    nested_lst.append(sub)

flat = []
for sub in nested_lst:
    for item in sub:
        flat.append(item)
print("Flattened list =", flat)
```

Q40. Find Palindrome Elements in a List

Q79. Find the Median of a List

Test Cases:

Input: [1, 2, 3, 4, 5] → Output: 3

 Input: [10, 20, 30, 40] → Output: 25.0

Q94. Find Armstrong Numbers in a List

Test Cases:

Input: [153, 370, 100] → Output: [153, 370]

 Input: [407, 200] → Output: [407]

 Input: [10, 20] → Output: []


```python
user_input = input("Enter positive integers separated by space: ")
lst = user_input.split()

armstrongs = []
for item in lst:
    num = int(item)
    power = len(item)
    total = 0
    for d in item:
        total += int(d) ** power
    if total == num:
        armstrongs.append(num)
print("Armstrong numbers =", armstrongs)
```

```python
lst = [153, 370, 100]
armstrongs = []
for num in lst:
    s = str(num)
    power = len(s)
    if sum(int(d) ** power for d in s) == num:
        armstrongs.append(num)
print(armstrongs)  # Output: [153, 370]
```

Q100. Find Anagram Pairs in a List

Test Cases:
