Q32. Find Duplicate Elements in a List

Test Cases:

Input: [1, 2, 2, 3, 3, 4] → Output: [2, 3]

 Input: ['a', 'b', 'a'] → Output: ['a']

 Input: [10, 20, 30] → Output: []


```python
lst = eval(input("Enter list: "))
seen = []
duplicates = []
i = 0
while i < len(lst):
    item = lst[i]
    if item in seen:
        if item not in duplicates:
            duplicates.append(item)
    else:
        seen.append(item)
    i += 1
print("Output:", duplicates)
```

Q35. Split a List into Chunks

Test Cases:

Input: [1, 2, 3, 4, 5, 6], size=2 → Output: [[1, 2], [3, 4], [5, 6]]

 Input: ['a', 'b', 'c', 'd'], size=2 → Output: [['a', 'b'], ['c', 'd']]

 Input: [10, 20, 30], size=1 → Output: [[10], [20], [30]]


```python
lst = eval(input("Enter list: "))
size = int(input("Enter chunk size: "))

result = []
i = 0

while i < len(lst):
    chunk = lst[i : i + size]
    result.append(chunk)
    i += size

print("Output:", result)
```

```python
lst = eval(input("Enter list: "))
size = int(input("Enter chunk size: "))
all_chunks = []
current_group = []

for item in lst:
    current_group.append(item)

    if len(current_group) == size:
        all_chunks.append(current_group)
        current_group = []

if len(current_group) > 0:
    all_chunks.append(current_group)

result = eval("all_chunks")
print("Output:", result)
```

Q51. Find All Prime Numbers in a List

Test Cases:

Input: [1, 2, 3, 4, 5, 6] → Output: [2, 3, 5]

 Input: [10, 11, 12, 13] → Output: [11, 13]

 Input: [4, 6, 8] → Output: []


```python
lst = eval(input("Enter list: "))
result = []
i = 0
while i < len(lst):
    num = lst[i]

    if num > 1:
        is_prime = True
        divisor = 2
        while divisor * divisor <= num:
            if num % divisor == 0:
                is_prime = False
            divisor += 1

        if is_prime:
            result.append(num)
    i += 1
print("Output:", result)
```

Q95. Find Perfect Numbers in a List

 Input: [496, 12] → Output: [496]

 Input: [5, 7] → Output: []


```python
lst = eval(input("Enter list: "))
result = []
i = 0

while i < len(lst):
    num = lst[i]

    if num > 1:
        divisor_sum = 0
        divisor = 1
        while divisor < num:
            if num % divisor == 0:
                divisor_sum += divisor
            divisor += 1

        if divisor_sum == num:
            result.append(num)
    i += 1
print("Output:", result)
```
