```python
num = int(input('Enter a number to reverse: '))

n = num # Init
rev = 0
while n > 0: # condition
  # Extract last digit
  ext_dig = n%10
  # Delete last digit after extraction
  n //= 10
  # Build reversed number
  rev = rev*10 + ext_dig
print('Reversed Number', rev)
```

To check whether the given number is palindrome or not

```python
num = int(input('Enter a number to reverse number: '))

n = num
rev = 0
while n > 0:
  ext_dig = n%10
  n //= 10
  rev = rev*10 + ext_dig
if num == rev:
  print('palindrome')
else:
  print('not palindrome')
```

To find sum of digits of a number

ex: 1234 -> 1+2+3+4 = 10

```python
num = int(input())
s = 0
n = num
while n > 0:
  ext_dig = n%10
  n //= 10
  s = s+ext_dig
print(s)
```
