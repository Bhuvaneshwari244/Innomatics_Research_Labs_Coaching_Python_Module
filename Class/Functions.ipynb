### *args and **kwargs - Variable length arguments

- *args - variable length positional arguments
- **kwargs - variable length keyword arguments

````python
def func(a, b):
  return f'a: {a}, b: {b}'  # f retuens the variable name
````

````python
func(5, 2)
````

````python
func(5, (1, 2, 3))
````

````python
func((1, 2))
````

````python
func(*(1, 2))
````

###variable length positional argument

````python
def func(a, b, *args):
  return f'a: {a}, b: {b}, args: {args}'
````

````python
func(1, 2, 3, 4, 5,6,7 ,8, 9,0)
````

This is called as `variable length positional arguments`

````python
func(tuple('variable length positional arguments'))
````

````python
func(*tuple('variable length positional arguments'))
````

###variable length keyword arguments

````python
def func(a, b, **kwargs):
  return f'a is {a}, b is {b}, kwargs: {kwargs}'
````

````python
func(a = 1, b = 2, c = 3, d = 4, e = 5, f = 6)
````
