````python
try:
  a = int(input('Enter a number: '))
  b = int(input('Enter a number:'))
  div = a/b
except (ZeroDivisionError, ValueError):
  print('ZeroDivisionError OR valueError')
print(f'The result of dividing {a} with {b} is {div}')
print('Success')
````

````python
try:
  a = int(input('Enter a number: '))
  b = int(input('Enter a number:'))
  div = a/b
except (ZeroDivisionError, ValueError) as err: # Created an object of Error
  print('Error is:', err)
print(f'The result of dividing {a} with {b} is {div}')
print('Success')
````

````python
try:
  a = int(input('Enter a number: '))
  b = int(input('Enter a number:'))
  div = a/b
except (ZeroDivisionError, ValueError) as err: # Created an object of Error
  print('Error is:', err)
else:
  print(f'The result of dividing {a} with {b} is {div}') # Statement is meaningful only when there is no exception
print('Success')
````
