## Iterator Protocol

Implementing 2 methods in the class, those methods are will be reflected in class directory
- `__iter__` - created an iterator
- `__next__` - which extract elements from container and datastructure


**For any data structure, if iterator protocol is implemented, then to access and extract elements, index are not required**

````python
'__iter__' in dir(list), '__next__' in dir(list)
````

To access and extract elements from data structure
1. Create an iterator
2. call `next` on it

By using while loop

````python
s = {1, 2.3, 'hello', False, None}
s
````

````python
set_iter = iter(s)
set_iter
````

````python
n = 0
while n < len(s):
    print(next(set_iter))
    n += 1
````

Following is the implementation of `for loop`

````python
s = {1, 2.3, 'hello', False, None}
s
````

````python
s = {1, 2.3, 'hello', False, None}
set_iter = iter(s)
n = 0
while n < len(s):
    print(next(set_iter))
    n += 1
````

In for loop
1. Creates an iterator object for the given `iterable`
2. Calls the next on iterator object till it raises an exception `StopIteration`

**iterable**
- Any datastructure in which `__iter__` method is implemented

````python
'__iter__' in dir(list) #  list is an iterable
````

````python
'__iter__' in dir(tuple) # tuple is an iterable
````
