Function vs Method

- Function
```
def func(param1, param2):
   LOGIC
   return output
```
- Method (BOUND METHOD) - BOUND TO THE Class/ Object
```
def method_name(reference of Class/Object (cls or self), param1, param2 ....):
    LOGIC
    return output
```
