List comprehension

```python
lst = [1, 2, 3, 4,5]

lst
```

```python
lst_output = []
for ele in lst:
  sq_ele = ele**2
  lst_output.append(sq_ele)
print(lst_output)
```

```python
lst_output = [ele**2 for ele in lst]
lst_output
```

syntax:

var = [ operation on ele      -      for ele in iterable       -  if < condition > on ele]

square only even number elements of an iterable

```python
lst = [1, 2, 3, 4, 5]
[ele**2  for ele in lst  if ele%2 == 0]
```

FILTER

```python
lst_out = [1, 2, 3, 4, 5]
lst_out = [j**2 for j in range(1, 11) if j%2 == 0]
lst_out
```

Dictionary comprehension

- frequently used application

```python
dct = {1:'a', 2:'b', 3:'c', 4:'d'}
dct
```

```python
for k, v in dct.items():
  print(f'k, v')
```

```python
dct.items()
```

1. k, v in (1, 'a')
2. k, v in (2, 'b')
3. k, v in (3, 'c')
4. k, v in (4, 'd')

```python
for k, v in dct.items():
  print(f'k is {k} and v is {v}')
```

```python
dct_out = {}
for k, v in dct.items():
  k, v = v, k
  dct_out[k] = v
dct_out
```

by using dct comprehension

```python
dct_out = {v:k for k, v in dct.items()}
dct_out
```

```python
# BUILT IN SCOPE - PRINT
a = 10 # GLOBAL
def outer():  # GLOBAL
  b = 20 # ENCLOSING
  print('a is', a)
  print('b is', b)

  def inner(): # ENCLOSING
    c = 30 # LOCAL
    print('a is', a)
    print('b is', b)
    print('c is', c)

  inner()
  return None
```

- `a` is defined in `global`

- `c` is defined in `local`
