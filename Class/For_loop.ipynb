```python
#while loop
a, b = 0, 1
while a < 100:
  print(a)
  a, b = b, a+b
```

```python
a, b = 0, 1
while True:
  print(a)
  a, b = b, a+b
  if b > 100:
    break
else:
  print('LOOP IS NOT TERNIMATED BECAUSE OF BREAK')
```

```python
a, b = 0, 1
print(a)
while True:
  a, b = b, a+b
  print(a)
  if b > 100:
    break
else:
  print('LOOP IS NOT TERNIMATED BECAUSE OF BREAK')
```

```python
a, b = 0, 1
print(a)
while True:
  a, b = b, a+b
  print(a)
  if b > 10000:
    break
else:
  print('LOOP IS NOT TERNIMATED BECAUSE OF BREAK')
```

```python
limit = int(input("Enter a limiting value for fib series : "))
a, b = 0, 1
print(a)
while True:
  a, b = b, a+b
  print(a)
  if b > limit:
    break
else:
  print('LOOP IS NOT TERNIMATED BECAUSE OF BREAK')
```
